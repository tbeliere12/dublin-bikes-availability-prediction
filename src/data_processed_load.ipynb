```python
import pandas as pd
# from pathlib import Path


all_bikes_path= "../../data/processed/dublin_bikes_all.csv"
df_bikes = pd.read_csv(all_bikes_path)

df_bikes.describe()
df_bikes.head(5)
df_bikes.tail(5)
```

```python
df_bikes.describe()
```

```python
df_bikes["bikes_gt_capacity"] = df_bikes["num_bikes_available"] > df_bikes["capacity"]
df_bikes["docks_gt_capacity"] = df_bikes["num_docks_available"] > df_bikes["capacity"]
df_bikes["matches_capacity"] = (
    df_bikes["num_bikes_available"] + df_bikes["num_docks_available"] == df_bikes["capacity"]
)
print("\nOverall:")
print(f"Bikes > cap: {df_bikes['bikes_gt_capacity'].sum()}")
print(f"Docks > cap: {df_bikes['docks_gt_capacity'].sum()}")

# Find the indices where docks_gt_capacity is True
docks_over_cap_indices = df_bikes.index[df_bikes["docks_gt_capacity"]].tolist()

# Print their locations (index and optionally other identifying columns)
print("\nRows where num_docks_available > capacity:")
print(df_bikes.loc[docks_over_cap_indices, ["station_id", "last_reported", "num_docks_available", "capacity"]])

# Replace num_docks_available with capacity for these rows
df_bikes.loc[docks_over_cap_indices, "num_docks_available"] = df_bikes.loc[docks_over_cap_indices, "capacity"]


violations = df_bikes[
    (df_bikes["bikes_gt_capacity"]) |
    (df_bikes["docks_gt_capacity"]) |
    (~df_bikes["matches_capacity"])
]

total_violations = len(violations)
print(f"Total number of violations: {total_violations}")


violations_by_station = violations.groupby("station_id").size().reset_index(name="violation_count")
violations_by_station.head(20)
```
